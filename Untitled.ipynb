```python
# Data libraries
import pandas as pd
import numpy as np

# Plotting libraries
import matplotlib.pyplot as plt
import seaborn as sns

# For creating seven-day windows
from numpy.lib.stride_tricks import sliding_window_view

# Load average river levels
df_avg = pd.read_csv("avg_river_levels.csv")
```

```python
# Preprocess river level data
def prepare_levels(df):

    # Select columns containing daily water levels
    date_cols = df.columns[3:]

    # Convert water levels two numerical values
    data = df[date_cols].apply(
        pd.to_numeric, errors="coerce"
    )

    # Replace invalid and extreme observations with NaN
    data = data.replace([np.inf, -np.inf], np.nan)
    data = data.mask(data.abs() > 40)

    # Convert column names to dates and sort chronologically
    data.columns = pd.to_datetime(date_cols, dayfirst=True)
    data = data.sort_index(axis=1)

    # Ensure every calendar date is represented
    all_dates = pd.date_range(
        start=data.columns.min(),
        end=data.columns.max(),
        freq="D"
    )

    data = data.reindex(columns=all_dates)

    # Exclude the first 908 days (very limited station coverage)
    data = data.iloc[:, 908:]

    # Transpose: rows = dates, columns = stations
    return data.T


# Prepare average river levels
avg = prepare_levels(df_avg)

display(avg.head())
print("Shape:", avg.shape)
```

WHY: we want river levels stored in chronological order, with one row per day. Missing days remain NaN to we dont accidentally treat observations several days apart as consecutive

```python
# Create seven-day windows for flood prediction
def valid_7day(df):

    # Extract river levels and dates
    values = df.to_numpy(dtype=float)
    dates = df.index

    # Create eight-day windows:
    # first seven days = predictors
    # eighth day = outcome
    periods = sliding_window_view(
        values, window_shape=8, axis=0
    )

    # Keep only windows containing eight valid observations
    valid = np.isfinite(periods).all(axis=2)

    # Identify valid dates and stations
    date_idx, station_idx = np.where(valid)

    # Extract valid windows
    valid_periods = periods[date_idx, station_idx, :]

    # Construct classification dataset
    classification_df = pd.DataFrame({
        "Station_ID": df.columns.to_numpy()[station_idx],
        "Week_Start": dates[date_idx],
        "Prediction_Date": dates[date_idx + 7],

        "Day_1": valid_periods[:, 0],
        "Day_2": valid_periods[:, 1],
        "Day_3": valid_periods[:, 2],
        "Day_4": valid_periods[:, 3],
        "Day_5": valid_periods[:, 4],
        "Day_6": valid_periods[:, 5],
        "Day_7": valid_periods[:, 6],

        "Next_Day": valid_periods[:, 7]
    })

    return classification_df


# Create the classification dataset
classification_df = valid_7day(avg)

display(classification_df.head())
print("Number of observations:", len(classification_df))
print("Missing values:", classification_df.isna().sum().sum())
```

```python
# Examine number of observations and flood events by year
classification_df["Year"] = classification_df["Prediction_Date"].dt.year

yearly_summary = classification_df.groupby("Year")["Flood"].agg(
    Observations="count",
    Flood_Events="sum",
    Flood_Rate="mean"
)

yearly_summary["Flood_Rate"] *= 100

display(yearly_summary)
```

```python
# Proposed flood definition: station-specific 99th percentile

# Use only training-period observations to determine thresholds
training_levels = avg.loc[avg.index < "2025-01-01"]

# Calculate the 99th percentile for each station
thresholds = training_levels.quantile(0.99)

# Require at least 365 valid training observations per station
training_counts = training_levels.notna().sum()
thresholds = thresholds.where(training_counts >= 365)

# Assign each observation its station-specific threshold
classification_df["Flood_Threshold"] = (
    classification_df["Station_ID"].map(thresholds)
)

# Exclude stations with insufficient training history
classification_df = classification_df.dropna(
    subset=["Flood_Threshold"]
).copy()

# Create binary response variable
classification_df["Flood"] = (
    classification_df["Next_Day"]
    > classification_df["Flood_Threshold"]
).astype(int)

# Check class distribution
print(classification_df["Flood"].value_counts())
print(classification_df["Flood"].value_counts(normalize=True) * 100)
```

investigating 99th percentile of historical water levels at each station, calculated using only training data. Could be changed to 95th if that is better. This defines an *unusually high water event* which is like kinda the same as flooding. NOT proof a flood has actually occured

ive also added 365 min observations minumum as calculated an extreme percentile from just a few days isnt really reliable. This is just a suggestion for data quality reasons.

```python
# Chronological train / validation / test split

# Training: before 2025
train = classification_df[
    classification_df["Prediction_Date"] < "2025-01-01"
].copy()

# Validation: 2025
val = classification_df[
    (classification_df["Prediction_Date"] >= "2025-01-01") &
    (classification_df["Prediction_Date"] < "2026-01-01")
].copy()

# Test: 2026 onwards
test = classification_df[
    classification_df["Prediction_Date"] >= "2026-01-01"
].copy()


# Check the number of observations and flood events
for name, data in [
    ("Train", train),
    ("Validation", val),
    ("Test", test)
]:
    print(f"\n{name}:")
    print("Observations:", len(data))
    print("Flood events:", data["Flood"].sum())
    print("Flood proportion:", data["Flood"].mean() * 100)
```
